```python
# University data dictionary with 10 key-value pairs
university = {
    'name': 'COMSATS University Islamabad',
    'location': 'Lahore Campus',
    'established': 2002,
    'students': 8000,
    'faculties': ['Engineering', 'IT', 'Business', 'Sciences'],
    'is_public': True,
    'rank': 3,
    'motto': 'In Search of Excellence',
    'website': 'https://lahore.comsats.edu.pk',
    'contact': {'email': 'info@comsats.edu.pk', 'phone': '042-111-001-007'}
}

# 1. .keys() - Get all keys in the dictionary
print("1. Keys:", university.keys())

# 2. .values() - Get all values
print("\n2. Values:", university.values())

# 3. .items() - Get all key-value pairs
print("\n3. Items:", university.items())

# 4. .get() - Get value for a specific key (safe way)
print("\n4. Motto of university:", university.get('motto'))
print("   Vice Chancellor:", university.get('vice_chancellor', 'Not available'))

# 5. .update() - Update or add multiple key-value pairs
university.update({'students': 8500, 'hostel_facility': True})
print("\n5. Updated university data:", university)

# 6. .pop() - Remove and return a key
removed = university.pop('rank', 'Not Found')
print("\n6. Removed rank:", removed)

# 7. .setdefault() - Add key with default if not exists
accreditation = university.setdefault('accredited_by', 'HEC')
print("\n7. Accreditation info:", accreditation)

# 8. .copy() - Create a shallow copy of the dictionary
university_copy = university.copy()
university_copy['name'] = 'COMSATS Abbottabad'
print("\n8. Original name:", university['name'], "| Copy name:", university_copy['name'])

# 9. len() - Get number of items in the dictionary
print("\n9. Total items in university dictionary:", len(university))

# 10. del - Delete a specific key
del university['is_public']
print("\n10. Dictionary after deleting 'is_public':", university)
```
